```python
aac = pd.read_csv('aac.csv')
dpc = pd.read_csv('dpc.csv')
pcp = pd.read_csv('pcp.csv')

featpos = pd.concat([aac, dpc, pcp], axis = 1)
```

```python
!pfeature_comp -i train_ne_cdhit.txt -j AAC -o aacneg.csv
```

```python
!pfeature_comp -i train_ne_cdhit.txt -j DPC -o dpcneg.csv
```

```python
aacneg = pd.read_csv('aacneg.csv')
dpcneg = pd.read_csv('dpcneg.csv')
pcpneg = pd.read_csv('pcpneg.csv')

featneg = pd.concat([aacneg, dpcneg, pcpneg], axis = 1)
```

```python
featneg['label'] = 'non-AMP'
```

```python
featneg.head()
```

```python
featneg.shape
```

```python
features = pd.concat([featpos, featneg], axis = 0)
```

```python
features.shape
```

```python
features.to_csv('features.csv', index = False)
```

Assigning X and Y

```python
x = features.drop('label', axis = 1)
y = features['label'].copy()
```

```python
y = y.map({'AMP' : 1, 'non-AMP': 0})
```

```python
y
```

Feature Selection

```python
from sklearn.feature_selection import VarianceThreshold

fs = VarianceThreshold(threshold = 0.1)
fs.fit_transform(x)

x2 = x.loc[:, fs.get_support()]
x2
```

Data splitting

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x2, y, test_size = 0.2, random_state = 42, stratify = y)
```

Model Building

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
clf = RandomForestClassifier(n_estimators = 500, random_state = 42)
clf.fit(x_train, y_train)
```

```python
y_pred = clf.predict(x_test)
```

```python
print('Accuracy:', clf.score(x_test, y_test))
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```
